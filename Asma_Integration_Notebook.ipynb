```python
# Entraînement de l'Arbre de Décision
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
import joblib

data = [[0, 10, 80], [0, 20, 50], [1, 10, 30], [1, 20, 40], [1, 50, 15], [1, 80, 10]]
df = pd.DataFrame(data, columns=['feu', 'vent', 'humidite'])
labels = [0, 0, 1, 2, 3, 3]

arbre = DecisionTreeClassifier()
arbre.fit(df, labels)
joblib.dump(arbre, "models/mon_arbre_alerte.pkl")
print("✅ Modèle de décision entraîné !")
```

```python
# Pipeline Complet (Salim + Mondir + Asma)
from predict import predict
from predict_propagation import predict_propagation
from predict_alert import donner_alerte

image = "dataset/satellite_samples/images/fire/sat_fire_0001.png"
res_salim = predict(image)
print(f"1. Salim (CNN) - Feu : {res_salim['fire_detected']}")

res_mondir = predict_propagation(location=res_salim["location"], wind={"speed": 38.0, "direction": "Sud-Ouest"}, humidity=22.0, temperature=34.0)
print(f"2. Mondir (LSTM) - Vitesse : {res_mondir['propagation_speed']} km/h")

res_asma = donner_alerte(1, 38.0, 22.0)
print(f"3. Asma (Décision) - ALERTE : {res_asma}")
```
